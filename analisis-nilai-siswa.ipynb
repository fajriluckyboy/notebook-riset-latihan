```python
# data nilai siswa, sengaja ditulis manual
# di sini untuk simulasi (nanti bisa
# dikembangkan baca langsung dari file CSV)
nilai_siswa = {
    "Budi Santoso": 85,
    "Siti Aminah": 90,
    "Rudi Hartono": 75,
    "Andi Wijaya": 80
}

# hitung rata-rata kelas
rata_rata = sum(nilai_siswa.values()) / len(nilai_siswa)
print(f"Rata-rata nilai kelas: {rata_rata:.2f}")
```
