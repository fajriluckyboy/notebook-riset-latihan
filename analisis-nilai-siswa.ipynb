# Analisis Nilai Siswa Minggu 1

Notebook ini menganalisis data nilai siswa
dari kelas Pengantar Git & GitHub.

```python
# data nilai siswa, sengaja ditulis manual
# di sini untuk simulasi (nanti bisa
# dikembangkan baca langsung dari file CSV)
nilai_siswa = {
    "Budi Santoso": 85,
    "Siti Aminah": 90,
    "Rudi Hartono": 75,
    "Andi Wijaya": 80
}

# hitung rata-rata kelas
rata_rata = sum(nilai_siswa.values()) / len(nilai_siswa)
print(f"Rata-rata nilai kelas: {rata_rata:.2f}")
```

```python
# tambahkan siswa baru ke data,
# lalu hitung ulang rata-rata kelas
nilai_siswa["Dewi Lestari"] = 88

rata_rata_baru = sum(nilai_siswa.values()) / len(nilai_siswa)
print(f"Rata-rata setelah tambah siswa baru: {rata_rata_baru:.2f}")
```
